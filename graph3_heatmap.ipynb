# Graph 3: Country vs. Maturity Rating Heatmap
**Library:** Altair  

```python
import altair as alt
import pandas as pd

df = pd.read_csv("graph3_data.csv")

# Derive sort orders dynamically
top_countries = df.groupby("country")["count"].sum().sort_values(ascending=False).index.tolist()
valid_ratings = ["G","TV-G","PG","TV-Y","TV-Y7","TV-PG","PG-13","TV-14","R","TV-MA","NR"]

metric_select = alt.binding_radio(
    options=["pct","count"],
    labels=["% of country","Raw count"],
    name="Show: ",
)
metric_param = alt.param(name="metric", bind=metric_select, value="pct")

base = (
    alt.Chart(df)
    .transform_calculate(value="metric == 'pct' ? datum.pct : datum.count")
    .mark_rect()
    .encode(
        x=alt.X("rating:N", sort=valid_ratings, title="Maturity Rating",
                axis=alt.Axis(labelAngle=-35, labelFontSize=11, titleFontSize=12)),
        y=alt.Y("country:N", sort=top_countries, title="Country",
                axis=alt.Axis(labelFontSize=11, titleFontSize=12)),
        color=alt.Color("value:Q", scale=alt.Scale(scheme="blues"),
                        legend=alt.Legend(title="Value")),
        tooltip=[
            alt.Tooltip("country:N", title="Country"),
            alt.Tooltip("rating:N",  title="Rating"),
            alt.Tooltip("count:Q",   title="Titles",          format=","),
            alt.Tooltip("pct:Q",     title="% of country",    format=".1f"),
        ],
    )
    .add_params(metric_param)
    .properties(
        width=580, height=320,
        title=alt.TitleParams(
            text="Country vs. Maturity Rating",
            subtitle="Toggle between % of country output and raw title count",
            fontSize=15, subtitleFontSize=11, subtitleColor="#666",
        ),
    )
)

text = (
    alt.Chart(df)
    .transform_calculate(value="metric == 'pct' ? datum.pct : datum.count")
    .transform_calculate(
        label="metric == 'pct' "
              "? (datum.pct > 8 ? format(datum.pct,'.0f') + '%' : '') "
              ": (datum.count > 0 ? datum.count : '')"
    )
    .mark_text(fontSize=9, color="white")
    .encode(
        x=alt.X("rating:N", sort=valid_ratings),
        y=alt.Y("country:N", sort=top_countries),
        text=alt.Text("label:N"),
        opacity=alt.condition(alt.datum.value > 0, alt.value(1), alt.value(0)),
    )
    .add_params(metric_param)
)

chart = (base + text).configure_view(strokeWidth=0)
chart.save("graph3_heatmap.html")
print("Saved -> graph3_heatmap.html")
chart
```
