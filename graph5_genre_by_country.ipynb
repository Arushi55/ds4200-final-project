# Graph 5: Genre Breakdown by Country

```python
import altair as alt
import pandas as pd
from genre_colors import GENRE_COLORS

df = pd.read_csv("graph5_data.csv")

# Derive country list and genre list dynamically from the data
top_countries = df.groupby("country")["count"].sum().sort_values(ascending=False).index.tolist()
top_genres    = df.groupby("genre")["count"].sum().sort_values(ascending=False).index.tolist()


# Dropdown to select country
country_select = alt.binding_select(options=top_countries, name="Country: ")
country_param  = alt.param(name="selected_country", bind=country_select, value=top_countries[0])

chart = (
    alt.Chart(df)
    .transform_filter("datum.country == selected_country")
    .mark_bar(cornerRadiusTopRight=4, cornerRadiusBottomRight=4)
    .encode(
        y=alt.Y(
            "genre:N",
            sort=alt.EncodingSortField(field="count", order="descending"),
            title="Genre"
        ),
        x=alt.X(
            "pct:Q",
            title="Percentage Of Country's Output"
        ),
        color=alt.Color(
            "genre:N",
            scale=alt.Scale(domain=list(GENRE_COLORS.keys()), range=list(GENRE_COLORS.values())),
            legend=None,
        ),
        tooltip=[
            alt.Tooltip("country:N", title="Country"),
            alt.Tooltip("genre:N",   title="Genre"),
            alt.Tooltip("count:Q",   title="Titles",            format=","),
            alt.Tooltip("pct:Q",     title="Percentage of Output",       format=".1f"),
        ],
    )
    .add_params(country_param)
    .properties(
        width=520,
        height=280,
        title=alt.TitleParams(
            text="Genre Breakdown by Country",
            fontSize=15,
        ),
    )
    .configure_view(strokeWidth=0)
    .configure_axis(grid=True, gridOpacity=0.3)
)

# Save as interactive HTML
chart.save("graph5_genre_by_country.html")
print("Saved → graph5_genre_by_country.html")

chart
```
