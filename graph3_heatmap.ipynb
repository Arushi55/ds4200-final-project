# Graph 3: Genre vs. Maturity Rating Heatmap

```python
import altair as alt
import pandas as pd

df = pd.read_csv("graph3_data.csv")

# Derive sort orders dynamically from the data
top_genres  = df.groupby("genre")["count"].sum().sort_values(ascending=False).index.tolist()
valid_ratings = ["G","TV-G","PG","TV-Y","TV-Y7","TV-PG","PG-13","TV-14","R","TV-MA","NR"]

# Radio button to toggle between % of genre and raw count
metric_select = alt.binding_radio(
    options=["pct", "count"],
    labels=["Percentage of Genre", "Raw Count"],
    name="Option: ",
)
metric_param = alt.param(name="metric", bind=metric_select, value="pct")

base = (
    alt.Chart(df)
    .transform_calculate(value="metric == 'pct' ? datum.pct : datum.count")
    .mark_rect()
    .encode(
        x=alt.X(
            "rating:N",
            sort=valid_ratings,
            title="Maturity Rating",
            axis=alt.Axis(labelAngle=360),
        ),
        y=alt.Y(
            "genre:N",
            sort=top_genres,
            title="Genre",
        ),
        color=alt.Color(
            "value:Q",
            scale=alt.Scale(scheme="blues"),
            legend=alt.Legend(title="Value"),
        ),
        tooltip=[
            alt.Tooltip("genre:N",  title="Genre"),
            alt.Tooltip("rating:N", title="Rating"),
            alt.Tooltip("count:Q",  title="Titles",     format=","),
            alt.Tooltip("pct:Q",    title="Percentage of genre", format=".1f"),
        ],
    )
    .add_params(metric_param)
    .properties(
        width=580,
        height=300,
        title=alt.TitleParams(
            text="Genre vs. Maturity Rating",
        ),
    )
)

text = (
    alt.Chart(df)
    .transform_calculate(value="metric == 'pct' ? datum.pct : datum.count")
    .transform_calculate(
        label="metric == 'pct' "
              "? (datum.pct > 5 ? format(datum.pct, '.0f') + '%' : '') "
              ": (datum.count > 0 ? datum.count : '')"
    )
    .mark_text(fontSize=9, color="white")
    .encode(
        x=alt.X("rating:N", sort=valid_ratings),
        y=alt.Y("genre:N",  sort=top_genres),
        text=alt.Text("label:N"),
        opacity=alt.condition(
            alt.datum.value > 0, alt.value(1), alt.value(0)
        ),
    )
    .add_params(metric_param)
)

chart = (base + text).configure_view(strokeWidth=0)
chart.save("graph3_heatmap.html")

chart
```
